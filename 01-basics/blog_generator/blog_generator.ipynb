```python
from google import genai
from dotenv import load_dotenv

load_dotenv()

client = genai.Client()
```

```python
from typing import TypedDict

class BlogState(TypedDict):

    blog_topic: str
    blog_outline: str
    final_blog: str
```

```python
def generate_outline(State: BlogState) -> BlogState:

    blog_topic = State['blog_topic']
    response = client.models.generate_content(
        model= "gemini-3.5-flash",
        contents= f"Generate an outline for the topic: {blog_topic}"
    )
    State['blog_outline'] = response.text

    return State
```

```python
def generate_blog(State: BlogState) -> BlogState:

    blog_outline = State['blog_outline']
    response = client.models.generate_content(
        model = "gemini-3.5-flash-lite",
        contents = f"Generate a blog on this outline: {blog_outline}"
    )
    State['final_blog'] = response.text

    return State
```

```python
from langgraph.graph import StateGraph, START, END

graph = StateGraph(BlogState)

graph.add_node('generate_outline', generate_outline)
graph.add_node('generate_blog', generate_blog)

graph.add_edge(START, 'generate_outline')
graph.add_edge('generate_outline', 'generate_blog')
graph.add_edge('generate_blog', END)

workflow = graph.compile()
```

```python
initial_state = {
    'blog_topic': "Problems in Today's Quantum Computers"
}

final_state = workflow.invoke(initial_state)

print(final_state['final_blog'])
```
